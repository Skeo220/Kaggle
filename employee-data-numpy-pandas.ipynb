```python
# function for average salary
def average_salary_by(category):
    avg_salary = df.groupby(category)['Salary'].mean().sort_values(ascending=False)
    return avg_salary

def max_salary_by(category):
    max_salary = df.groupby(category)['Salary'].max().sort_values(ascending=False)
    return max_salary

def min_salary_by(category):
    min_salary = df.groupby(category)['Salary'].min().sort_values(ascending=False)
    return min_salary

def total_salary_by(category):
    total_salary= df.groupby(category)['Salary'].sum().sort_values(ascending=False)
    return total_salary
```

```python
def generate_bins_for(column, bin_edge_min, bin_edge_max, bin_interval):
    """
        takes in a numerical column and creates bins based on specified parameters.
        
        parameters:

            column = the dataframe column to be binned. Must be a numerical column
            bin_edge_min = starting bin edge value
            bin_edge_max = ending bin edge value. Must be greater than the bin edge min by at least 1 interval
            bin_interval = determines the exact bin edges between the bin_edge_min and bin_edge_max

        returns:
            a array like object representing the bin for each value -> pd.cut 
    """

    if df[column].dtype != int:
        print("Column Type Error. Must provide a numerical column")
    else:
        if bin_edge_max >= bin_edge_min + bin_interval:
            bin_array = np.arange(bin_edge_min, bin_edge_max, bin_interval)
            binned_data = pd.cut(df[column], bins = bin_array, precision=0, right=False)
            return binned_data
        else:
            print("bin_edge_max must be greater than bin_edge_min by at least 1 interval")
```

```python
df['age_group'] = generate_bins_for("Age",20,65,5)
df['salary_group'] = generate_bins_for("Salary", 0, 250000, 25000)
```
