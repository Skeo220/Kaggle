 Average pay by age, gender, title, experience, education, and salary

```python
# function for average salary
def average_salary_by(category):
    avg_salary = df.groupby(category)['Salary'].mean().sort_values(ascending=False)
    return avg_salary

def max_salary_by(category):
    max_salary = df.groupby(category)['Salary'].max().sort_values(ascending=False)
    return max_salary

def min_salary_by(category):
    min_salary = df.groupby(category)['Salary'].min().sort_values(ascending=False)
    return min_salary

def total_salary_by(category):
    total_salary= df.groupby(category)['Salary'].sum().sort_values(ascending=False)
    return total_salary
```

Start with getting the upper bound.

To get the upper bound, you have to take the lower bound, increment it by an interval until the upper bound is greater or equal to the max value.

```python
bins = []

for i in range(df['Age'].min()-1, df['Age'].max()):
    bins.append((i++ 5))
```

```python
def get_lower_bound(column):
    """
        get the lower bound of a numerical series.
        returns an even number.
    """
    if df[column].dtype == int:
        if df[column].min() % 2 != 0:
            lower_bound = df[column].min() - 1
            print(lower_bound)
    else: print("Column type error. Must be of type int.")
```
